```python
plt.figure(figsize=(8,5))
sns.boxplot(data=wine_data, x="wine_type", y="alcohol")
plt.title("Alcohol Content: Red vs White Wine")
plt.xlabel("Wine Type")
plt.ylabel("Alcohol")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=wine_data, x="wine_type", y="pH", )

plt.title("pH: Red vs White Wine")
plt.xlabel("Wine Type")
plt.ylabel("pH")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=wine_data, x="wine_type", y="residual sugar")

plt.title("Residual Sugar: Red vs White Wine")
plt.xlabel("Wine Type")
plt.ylabel("Residual Sugar")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=wine_data, x="wine_type", y="volatile acidity", color= 'slateblue')
plt.title("Volatile Acidity: Red vs White Wine")
plt.xlabel("Wine Type")
plt.ylabel("Volatile Acidity")
plt.show()
```
